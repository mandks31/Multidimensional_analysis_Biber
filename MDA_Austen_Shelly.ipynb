## Comparing Austen & Shelley Using Multidimensional Analysis 

Hey there! This notebook runs a quick stylistic comparison between Jane Austen’s *Pride and Prejudice* and Mary Shelley’s *Frankenstein* using Doug Biber’s multidimensional analysis.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sn
```

First, we'll load both datasets, merge them into a single DataFrame, and visualize their Z-scores across Biber's 6 dimensions. 


```python
df_austen = pd.read_csv("Dimensions_jane-austen-pride-prejudice.csv")
df_frankenstein = pd.read_csv("Dimensions_Frankenstein.csv")

df_combined = pd.concat([df_austen, df_frankenstein], ignore_index=True)
df_combined['Text'] = ['Pride and Prejudice', 'Frankenstein']

heatmap_data = df_combined.set_index("Text")[
    [
        "Dimension1",
        "Dimension2",
        "Dimension3",
        "Dimension4",
        "Dimension5",
        "Dimension6",
    ]
]
# 3. Set up the plotting canvas
plt.figure(figsize=(10, 4))

sns.heatmap(
    heatmap_data,
    annot=True,
    cmap="Purples",
    fmt=".2f",
    linewidths=0.5,
    cbar_kws={"label": "Z-Score Value"},
)

# 5. Format labels and title
plt.title(
    "Multidimensional Analysis: Austen x Shelly Dimension Scores Heatmap",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Biber Dimensions", fontsize=12)
plt.ylabel("Texts", fontsize=12)

# 6. Display the plot
plt.tight_layout()
plt.show()
```

#### Results

Looking at the heatmap, a couple of patterns stand out. Here's a possible interpretation:

* **Dimension 1 (Involved vs. Informational):** *Pride and Prejudice* scored high (+3.82), which makes total sense given all the interpersonal dialogue and character chatter. *Frankenstein* goes negative (-3.37), leaning way more into a distanced, narrative storytelling vibe.
* **Dimension 2 (Narrative Focus):** Both books are quite heavy on past-tense verbs and storytelling framing, but Austen peaks even higher here (+5.10).
* **Dimension 3 (Elaborated Reference):** *Frankenstein* dominates this category (+4.92), reflecting that heavy Gothic/Romantic style with nested frames (a story within a story within a story). and complex structural details.
* **Dimension 4 (Persuasion & Stance):** Both show solid positive scores, pointing to plenty of social commentary, opinions, and moral evaluations going on in both worlds.
